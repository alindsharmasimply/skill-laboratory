```python
import torch
```

```python
# Define a PyTorch tensor with shape (3,4) containing integers
# from 1 to 12 sequentially. Get creative to define it as cleanly as
# possible.
a = torch.arange(1, 13).reshape(3, 4)
print(a)
```

```python
# Change the dtype of the above tensor to float32.
print(f"Current type of tensor 'a': {a.dtype}")
a = a.to(torch.float32)
print(f"Changed type of tensor 'a': {a.dtype}")
```

```python
# Transpose its rows and columns (no need to change the input,
# just return a view).
print(f"Current tensor: {a}")
print(f"Transpose of the tensor: {a.transpose(dim0=0, dim1=1)}")
```

```python
# [Masking] Given a PyTorch tensor containing negative and positive integers, replace
# all the negative integers with -1.
b = torch.tensor([[1, -2, 3], [-4, -3, 0]])
print(b)
neg_mask = b < 0
print(neg_mask)
b[neg_mask] = -1
print(b)
```

```python
# Given 2D PyTorch tensors a and b, defined as
#   a = torch.tensor([[-3,1,2],[-1,0,6]])
#   b = torch.tensor([[0,1,0],[1,0,1]])
# For each of the tensors, do the following:
# 1. Check if there is any occurrence of the values 0 or 1.
# 2. Check if all the values of the tensor belong to the set 0, 1.
a = torch.tensor([[-3,1,2],[-1,0,6]])
b = torch.tensor([[0,1,0],[1,0,1]])
c = torch.tensor([0, 1])
torch.isin(a, c)

# Part-1
print(torch.isin(a, c).any().item())
print(torch.isin(b, c).any().item())

# Part-2
print(torch.isin(a, c).all().item())
print(torch.isin(b, c).all().item())
```

```python
# [Clipping] Given a 2D PyTorch tensor,
# 1. Return a tensor with all its elements clipped between 0 and 1.
#   So all the elements less than 0 should be changed to 0, all those
#   greater than 1 should be changed to 1, and those between 0 and 1
#   should be left unchanged.
# 2. Return a clipping of only its first row between 0 and 1.
# 3. Change the input tensor itself, clipping its first row between 0 and 1.
# You can start with the following input tensor:
#   a = torch.tensor([[-1, 0, 0.2, 1, 1.5],[2, -2, 0, 1, 0.4]])

a = torch.tensor([[-1, 0, 0.2, 1, 1.5],[2, -2, 0, 1, 0.4]])

# Part-1
print(f"clamp: {a.clamp(min=0, max=1)}")
print(f"clip: {a.clip(min=0, max=1)}") # Just an alias for clamp()
print(f"clamp: {torch.clamp(a, min=0, max=1)}")

# Part-2
print(f"{torch.clamp(a[0, :], min=0, max=1)}")
print(f"{a}")

# Part-3
a[0, :].clamp_(min=0, max=1) # The underscore(_) suffix is the in-place version for the corresponding clamp operation
print(f"{a}")
```

```python
# In PyTorch, image datasets are stored in the format (N, C, H, W), where N is the
# number of images in the set (or the batch), C is the number of channels (usually 3 for
# a color image, corresponding to red, green, and blue), H is the image height, and W
# is the image width. Given a tensor containing an image dataset of size [128, 3, 224,
# 224], perform the following slicing operations:
# 1. Select the first 32 images of the set.
# 2. Select only channel 0 for all the samples in the set (note
# that the output size must be [128, 1, 224, 224], and not [128,
# 224, 224]).
# 3. For each image in the set, and for all the channels, select pixels
# with height from 16 to 128 (inclusive of 16, but not 128), and
# select every alternate pixel from the width, starting from the
# beginning.
# You can work with a random tensor initialized as follows:
#   image_set = torch.rand([128, 3, 224, 224])

image_set = torch.rand([128, 3, 224, 224])

# Part-1:
print(image_set[:32, :, :, :].size())

# Part-2:
print(image_set[:, :1, :, :].size())

# Part-3:
print(image_set[:, :, 16:128, ::2].size())
```

```python
# Given a 2D tensor, calculate the index of the maximum element in every row. Repeat
# the same for each column.
# HINT: Use torch.argmax function.
# You can start with the following input tensor to test your code:
#   input_tensor = torch.tensor([[1,2,3,4,5,3],[3,5,6,3,2,1]])

input_tensor = torch.tensor([[1,2,3,4,5,3],[3,5,6,3,2,1]])

print(f"Row-wise max: {input_tensor.argmax(dim=1)}")

print(f"Column-wise max: {input_tensor.argmax(dim=0)}")
```

```python
# Given two 1D tensors of the same length, calculate the number of indices where they
# have the same values.
# You can use the following two input tensors to test your code:
#   target = torch.tensor([1,3,4,5,6,7])
#   predictions = torch.tensor([2,3,4,6,3,7])

target = torch.tensor([1,3,4,5,6,7])
predictions = torch.tensor([2,3,4,6,3,7])

print((target == predictions))
print((target == predictions).long())
print((target == predictions).long().sum())
print((target == predictions).long().sum().item())
```
